```python
import sqlite3
import pandas as pd

# Connect to Airtel database
con = sqlite3.connect(
    r"C:\Users\jumab\OneDrive\Documents\airtel-telecom-analytics\data\raw\airtel.db"
)

def q(sql):
    return pd.read_sql(sql, con)
```

```python
tables = q("""
SELECT name
FROM sqlite_master
WHERE type = 'table'
ORDER BY name
""")

display(tables)
```

```python
for table in tables["name"]:
    count = q(f'SELECT COUNT(*) AS total FROM "{table}"')
    print(f"{table}: {count.iloc[0, 0]} rows")
```

```python
counts = q("""
SELECT 'circles' AS tbl, COUNT(*) AS n FROM circles
UNION ALL
SELECT 'plans', COUNT(*) FROM plans
UNION ALL
SELECT 'customers', COUNT(*) FROM customers
UNION ALL
SELECT 'usage_daily', COUNT(*) FROM usage_daily
UNION ALL
SELECT 'recharges', COUNT(*) FROM recharges
UNION ALL
SELECT 'complaints', COUNT(*) FROM complaints
ORDER BY tbl;
""")

display(counts)
```
